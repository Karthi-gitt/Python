```python
#student marks analysis

students = ['A', 'B', 'C', 'D', 'E', 'F', 'G', 'H', 'I', 'J']
marks = [78, 85, 92, 67, 88, 95, 72, 81, 90, 76]

# Task 1: Display student marks graphically
plt.figure(figsize=(8, 5))
plt.bar(students, marks, color='skyblue')
```

```python
# Task 2: Find the highest scorer
highest_mark = max(marks)
highest_student = students[marks.index(highest_mark)]

# Highlight the highest scorer
plt.bar(highest_student, highest_mark, color='yellow', label='Highest Scorer')

#Task 3: Add title and axis labels
plt.title("Student Marks Analysis")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.legend()

plt.show()

# Print highest scorer
print("Highest Scorer:", highest_student)
print("Highest Marks:", highest_mark)
```

```python
# Task 4: Display marks in descending order and plot again
sorted_data = sorted(zip(students, marks), key=lambda x: x[1], reverse=True)

sorted_students = [x[0] for x in sorted_data]
sorted_marks = [x[1] for x in sorted_data]

print("\nMarks in Descending Order:")
for s, m in sorted_data:
  print(s, ":", m)
```

```python
# Plot sorted marks
plt.figure(figsize=(8, 5))
plt.bar(sorted_students, sorted_marks, color='green')
plt.title("Student Marks (Descending Order)")
plt.xlabel("Students")
plt.ylabel("Marks")

plt.show()
```
