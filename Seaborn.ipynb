```python
#Seaborn Example
import seaborn as sns
import pandas as pd

df = pd.DataFrame({
    'Students':['A','B','C','D'],
    'Marks':[78,85,92,88]
})

sns.barplot(x='Students', y='Marks', data=df)
plt.title("Student Marks")
plt.show()
```

```python
# Load dataset
tips = sns.load_dataset("tips")
```

```python
# 1. Scatter Plot
plt.figure(figsize=(6,4))
sns.scatterplot(x="total_bill", y="tip", data=tips, hue="sex", style="time")
plt.title("Scatter Plot - Total Bill vs Tip")
plt.show()
```

```python
# 2. Line Plot
plt.figure(figsize=(6,4))
sns.lineplot(x="size", y="total_bill", data=tips, ci=None)
plt.title("Line Plot - Total Bill vs Size")
plt.show()
```

```python
# 3. Bar Plot
plt.figure(figsize=(6,4))
sns.barplot(x="day", y="total_bill", data=tips, hue="sex")
plt.title("Bar Plot - Average Total Bill per Day")
plt.show()
```

```python
# 4. Count Plot
plt.figure(figsize=(6,4))
sns.countplot(x="day", data=tips, hue="sex")
plt.title("Count Plot - Customers per Day")
plt.show()
```

```python
# 5. Box Plot
plt.figure(figsize=(6,4))
sns.boxplot(x="day", y="total_bill", data=tips, hue="sex")
plt.title("Box Plot - Total Bill Distribution")
plt.show()
```

```python
# 6. Violin Plot
plt.figure(figsize=(6,4))
sns.violinplot(x="day", y="total_bill", data=tips, hue="sex", split=True)
plt.title("Violin Plot - Distribution by Day and Gender")
plt.show()
```

```python
# 7. Heatmap (Correlation Matrix)
plt.figure(figsize=(6,4))
corr = tips.corr(numeric_only=True)
sns.heatmap(corr, annot=True, cmap="coolwarm", linewidths=0.5)
plt.title("Correlation Heatmap")
plt.show()
```

```python
 #8. Pairplot
sns.pairplot(tips, hue="sex")
plt.suptitle("Pairplot of Tips Dataset",y=1.02)
plt.show()
```

```python
# 9. Distribution Plot (Histogram + KDE)
plt.figure(figsize=(6,4))
sns.histplot(tips["total_bill"], kde=True, bins=20)
plt.title("Distribution of Total Bill")
plt.show()
```

```python
# 10. Regression Plot
plt.figure(figsize=(6,4))
sns.regplot(x="total_bill", y="tip", data=tips)
plt.title("Regression Plot - Tip vs Total Bill")
plt.show()
```

```python
#Task Iris Data Visualization
iris = sns.load_dataset("iris")
```

```python
#1. Display Dataset Information

print("First 5 Rows of the Dataset:")
print(iris.head())
```

```python
print("\nDataset Information:")
print(iris.info())
```

```python
print("\nStatistical Summary:")
print(iris.describe())
```

```python
# 2. Scatter plot between petal length and petal width
plt.figure(figsize=(8,6))
sns.scatterplot(
    data=iris,
    x="petal_length",
    y="petal_width",
    hue="species",
    palette="Set1",
    s=80
)

plt.title("Petal Length vs Petal Width")
plt.xlabel("Petal Length (cm)")
plt.ylabel("Petal Width (cm)")
plt.legend(title="Species")
plt.show()
```

```python
# 3. Pair Plot
sns.pairplot(iris, hue="species", palette="Set1")
plt.show()
```

```python
# 4. Observation
print("\nObservation:")
print("The species 'setosa' appears to be the most distinct because")
print("its petal length and petal width are clearly separated from")
print("the other two species (versicolor and virginica).")
print("Versicolor and Virginica have some overlap in their measurements.")
```
