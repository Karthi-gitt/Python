```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("weight-height.csv")
```

```python
# Display first few rows
print("Dataset:")
print(df.head())
```

```python
# (1) Plot histograms for Height and Weight


plt.figure(figsize=(12,5))

plt.subplot(1,2,1)
plt.hist(df["Height"], bins=10, edgecolor='black')
plt.title("Height Histogram")
plt.xlabel("Height")
plt.ylabel("Frequency")

plt.subplot(1,2,2)
plt.hist(df["Weight"], bins=10, edgecolor='black')
plt.title("Weight Histogram")
plt.xlabel("Weight")
plt.ylabel("Frequency")

plt.tight_layout()
plt.show()
```

```python
# (2) Detect Weight Outliers using IQR


Q1 = df["Weight"].quantile(0.25)
Q3 = df["Weight"].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

weight_outliers = df[(df["Weight"] < lower) | (df["Weight"] > upper)]

print("\nWeight Outliers:")
print(weight_outliers)

# (3) Detect Height Outliers using IQR
# ---------------------------------------------------

Q1 = df["Height"].quantile(0.25)
Q3 = df["Height"].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

height_outliers = df[(df["Height"] < lower) | (df["Height"] > upper)]

print("\nHeight Outliers:")
print(height_outliers)
```

```python
# (4) Replace Outliers with Mean


df_mean = df.copy()

# Weight
mean_weight = df_mean["Weight"].mean()

Q1 = df_mean["Weight"].quantile(0.25)
Q3 = df_mean["Weight"].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

df_mean.loc[(df_mean["Weight"] < lower) | (df_mean["Weight"] > upper), "Weight"] = mean_weight

# Height
mean_height = df_mean["Height"].mean()

Q1 = df_mean["Height"].quantile(0.25)
Q3 = df_mean["Height"].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

df_mean.loc[(df_mean["Height"] < lower) | (df_mean["Height"] > upper), "Height"] = mean_height

print("\nDataset after replacing outliers with Mean:")
print(df_mean)

# ---------------------------------------------------
# Replace Outliers with Median
# ---------------------------------------------------

df_median = df.copy()

# Weight
median_weight = df_median["Weight"].median()

Q1 = df_median["Weight"].quantile(0.25)
Q3 = df_median["Weight"].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

df_median.loc[(df_median["Weight"] < lower) | (df_median["Weight"] > upper), "Weight"] = median_weight

# Height
median_height = df_median["Height"].median()

Q1 = df_median["Height"].quantile(0.25)
Q3 = df_median["Height"].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

df_median.loc[(df_median["Height"] < lower) | (df_median["Height"] > upper), "Height"] = median_height

print("\nDataset after replacing outliers with Median:")
print(df_median)
```
