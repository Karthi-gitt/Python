```python
#create a numpy array containing 1 to 20 numbers using inbuilt function
import numpy as np
arr=np.arange(1,21)
print(arr)
```

```python
#create an array of all zeros with 10 elements and convert it to matrix
import numpy as np
arr=np.zeros(10)
print(arr)
arr=arr.reshape(2,5)
print(arr)
```

```python
#array containing even numbers from 2 to 20
import numpy as np
even=np.arange(2,21,2)
print(even)
```

```python
#finding dimensions,shape,size,datatype and multiply every element by 2
import numpy as np
arr=np.array([[1,2,3],[4,5,6]])
print("Array:")
print(arr)
print("Dimensions:",arr.ndim)
print("Shape:",arr.shape)
print("Size:",arr.size)
print("Datatype:",arr.dtype)
print("Multiply by 2:")
print(arr*2)
```

```python
#find sum,mean,max,min
import numpy as np
arr=np.array([10,20,30,40,50])
print("Array:",arr)
print("Sum:",np.sum(arr))
print("Mean:",np.mean(arr))
print("Max:",np.max(arr))
print("Min:",np.min(arr))
```
