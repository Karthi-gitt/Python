```python
import numpy as np
import pandas as pd
```

```python
#create series
s = pd.Series([1,2,3,4,5,6,7])
print(s)
```

```python
# Descriptive statistics
print(s.describe())
```

```python
print(s.mean())
print(s.std())
print(s.min())
print(s.max())
```

```python
mapping = s.map(lambda x :x*2)
print(mapping)
print(s)
```

```python
applies = s.apply(lambda x : x ** 2)
print(applies)
print(s)
```

```python
s1 = pd.Series([4,7,54,23,92,41])
sorted = s1.sort_values()
print(sorted)
```

```python
dropped = s.drop(1)
print(dropped)
print(s)
```

```python
#Handling Missing data
s2 = pd.Series([2,4,6,np.nan,7,0])
print(s2.isnull())
```

```python
#opposite of isnull()
print(s2.notnull())
```

```python
#Filling missing values
print(s2.fillna(0))
```

```python
s3 = pd.Series([4,7,3,None,8,2,np.nan])
print(s3)
print(s3.dropna())
```

```python
print(s.iloc[2])  #1,2,3,4,5,6,7
print(s.iloc[-2])  #6
```

```python
greater_num = s[s<=5]
print(greater_num)
```

```python
print(s.sum())
```

```python
print(s.aggregate(['sum','std','mean']))
```
